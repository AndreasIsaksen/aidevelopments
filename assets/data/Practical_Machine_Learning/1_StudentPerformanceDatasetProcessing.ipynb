### Custom Functions for Analysis
To facilitate the analysis of the dataset, we'll define some custom functions that can help us visualize and understand the data better.

```python
def get_df_name(df):
    name =[x for x in globals() if globals()[x] is df][0]
    return name

def corr_matrix(df):
    numCorr1 = df.corr()
    fig, ax = plt.subplots(1,1, figsize=(23,7))
    sb.heatmap(numCorr1, annot=True, cmap='coolwarm', ax=ax)
    ax.set_title(get_df_name(df))

def target_bar(df1):
    fig, ax = plt.subplots(1, 1, figsize=(10, 5))
    sb.countplot(x="Target", data=df1, ax=ax)
    ax.set_title(get_df_name(df1))
    ax.set_xlabel("Target")
    ax.set_ylabel("Count")
    plt.tight_layout()
```

#### Numerical Feature Adjustment
By evaluating the distributions of the features with heatmaps and boxplots. The following adjustments will be made:
- `Curricular units 1st sem (credited)` and `Curricular units 2nd sem (credited)` - With a corr elation of 0.94, these two features are highly correlated. We can consider combining them into a single feature representing the total credited units across both semesters.
- `Curricular units 1st sem (enrolled)` and `Curricular units 2nd sem (enrolled)` - With a correlation of 0.94, these features these two features are highly correlated. We can consider combining them into a single feature representing the total enrolled units across both semesters.
- `Curricular units 1st sem (evaluations)` and `Curricular units 2nd sem (evaluations)` - With a correlation of 0.78, these features these two features are highly correlated. We can consider combining them into a single feature representing the total evaluations across both semesters.
- `Curricular units 1st sem (approved)` and `Curricular units 2nd sem (approved)` - With a correlation of 0.90, these features these two features are highly correlated. We can consider combining them into a single feature representing the total approved units across both semesters.
- `Curricular units 1st sem (grade)` and `Curricular units 2nd sem (grade)` - With a correlation of 0.84, these features these two features are highly correlated. We can consider combining them into a single feature representing the average grade across both semesters.
- `Mother's occupation` and `Father's occupation` - With a correlation of 0.91, these two features are highly correlated. We might consider merging these two features or creating a new feature that captures the combined effect of parental occupation on student performance.
- `International` and `Nationality` - With a correlation of 0.79, these two features are correlated. Since `International` has very low variance, we can consider dropping this feature and relying on `Nationality` to capture any relevant information about international students.

It's worth mentioning that there are also observed numerical features with a lot of outliers, such as `Course` and `Previous qualifications`. These features may require further investigation and potential outlier handling techniques to ensure they do not adversely affect model performance.

```python
# Combine `Curricular units 1st sem (credited)` and `Curricular units 2nd sem (credited)`
for col1, col2 in [('Curricular units 1st sem (credited)', 'Curricular units 2nd sem (credited)')]:
    student_numerical['Total credited units'] = (student_numerical[col1] + student_numerical[col2]) / 2
    student_numerical = student_numerical.drop(columns=[col1, col2])

# Combine `Curricular units 1st sem (enrolled)` and `Curricular units 2nd sem (enrolled)`
for col1, col2 in [('Curricular units 1st sem (enrolled)', 'Curricular units 2nd sem (enrolled)')]:
    student_numerical['Total enrolled units'] = (student_numerical[col1] + student_numerical[col2]) / 2
    student_numerical = student_numerical.drop(columns=[col1, col2])

# Combine `Curricular units 1st sem (evaluations)` and `Curricular units 2nd sem (evaluations)`
for col1, col2 in [('Curricular units 1st sem (evaluations)', 'Curricular units 2nd sem (evaluations)')]:
    student_numerical['Total evaluations'] = (student_numerical[col1] + student_numerical[col2]) / 2
    student_numerical = student_numerical.drop(columns=[col1, col2])

# Combine `Curricular units 1st sem (approved)` and `Curricular units 2nd sem (approved)`
for col1, col2 in [('Curricular units 1st sem (approved)', 'Curricular units 2nd sem (approved)')]:
    student_numerical['Total approved units'] = (student_numerical[col1] + student_numerical[col2]) / 2
    student_numerical = student_numerical.drop(columns=[col1, col2])

# Combine `Curricular units 1st sem (grade)` and `Curricular units 2nd sem (grade)`
for col1, col2 in [('Curricular units 1st sem (grade)', 'Curricular units 2nd sem (grade)')]:
    student_numerical['Average grade'] = (student_numerical[col1] + student_numerical[col2]) / 2
    student_numerical = student_numerical.drop(columns=[col1, col2])

# Combine `Mother's occupation` and `Father's occupation`
for col1, col2 in [("Mother's occupation", "Father's occupation")]:
    student_numerical['Combined parental occupation'] = student_numerical[col1] + student_numerical[col2] / 2
    student_numerical = student_numerical.drop(columns=[col1, col2])
```
